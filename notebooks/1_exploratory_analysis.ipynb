```python

def generate_dataset(experiments, doses, times, images_per_condition, cells_per_image):

    records = []
    rng = np.random.default_rng(42)

    time_factors = {
        0.5: 0.3,
        2: 0.7,
        24: 1.0
    }

    for experiment in experiments:
        experiment_variation = rng.normal(0, 50)

        for dose in doses:
            for time in times:

                treatment_effect = -5 * dose * time_factors[time]

                for image in range(1, images_per_condition + 1):
                    image_variation = rng.normal(0, 30)

                    for cell in range(1, cells_per_image + 1):
                        cell_variation = rng.normal(0, 80)

                        ad_tau = (
                            1500
                            + experiment_variation
                            + image_variation
                            + cell_variation
                            + treatment_effect
                        )

                        records.append({
                            "Experiment": experiment,
                            "Dose": dose,
                            "Time": time,
                            "Image": image,
                            "Cell": cell,
                            "Lifetime": ad_tau
                        })

    return pd.DataFrame(records)
```

```python

from scipy import stats

selected_time = 24
selected_dose = 20

# 1. Naive cell-level comparison
control_cells = df.loc[
    (df["Time"] == selected_time) &
    (df["Dose"] == 0),
    "Lifetime"
]

treated_cells = df.loc[
    (df["Time"] == selected_time) &
    (df["Dose"] == selected_dose),
    "Lifetime"
]

naive_test = stats.ttest_ind(
    treated_cells,
    control_cells,
    equal_var=False
)

# 2. Experiment-level paired comparison
paired_data = (
    experiment_df[
        (experiment_df["Time"] == selected_time) &
        (experiment_df["Dose"].isin([0, selected_dose]))
    ]
    .pivot(
        index="Experiment",
        columns="Dose",
        values="Mean_Lifetime"
    )
)

paired_test = stats.ttest_rel(
    paired_data[selected_dose],
    paired_data[0]
)

results = pd.DataFrame({
    "Method": [
        "Naive cell-level Welch",
        "Experiment-level paired"
    ],
    "T_statistic": [
        naive_test.statistic,
        paired_test.statistic
    ],
    "P_value": [
        naive_test.pvalue,
        paired_test.pvalue
    ]
})

display(results)
display(paired_data)
```

```python

# Treatment minus control within each experiment
paired_differences = (
    paired_data[selected_dose] - paired_data[0]
)

# Summary statistics
n = len(paired_differences)
mean_difference = paired_differences.mean()
sd_difference = paired_differences.std(ddof=1)
se_difference = stats.sem(paired_differences)

# 95% confidence interval using Student's t distribution
confidence_level = 0.95

ci_lower, ci_upper = stats.t.interval(
    confidence=confidence_level,
    df=n - 1,
    loc=mean_difference,
    scale=se_difference
)

print(f"Number of experiments: {n}")
print(f"Mean treatment effect: {mean_difference:.2f} ps")
print(f"SD of paired differences: {sd_difference:.2f} ps")
print(f"SE of paired differences: {se_difference:.2f} ps")
print(f"95% CI: [{ci_lower:.2f}, {ci_upper:.2f}] ps")
```

```python

import statsmodels.formula.api as smf

# Select the two conditions
model_df = image_df[
    (image_df["Time"] == 24) &
    (image_df["Dose"].isin([0, 20]))
].copy()

# Binary treatment indicator
model_df["Treatment"] = (model_df["Dose"] == 20).astype(int)

# Fit a mixed-effects model
model = smf.mixedlm(
    formula="Mean_Lifetime ~ Treatment",
    data=model_df,
    groups=model_df["Experiment"]
)

result = model.fit(reml=True)

print(result.summary())
```

```python

# Use all image-level observations
full_model_df = image_df.copy()

# Treat dose and time as categorical variables
full_model_df["Dose_cat"] = full_model_df["Dose"].astype("category")
full_model_df["Time_cat"] = full_model_df["Time"].astype("category")

# Fit the full model with dose × time interaction
full_model = smf.mixedlm(
    formula="Mean_Lifetime ~ C(Dose_cat) * C(Time_cat)",
    data=full_model_df,
    groups=full_model_df["Experiment"]
)

full_result = full_model.fit(reml=True)

print(full_result.summary())
```

```python

# Arrange experiment-level means by dose
paired_means = experiment_df.pivot(
    index=["Experiment", "Time"],
    columns="Dose",
    values="Mean_Lifetime"
)

# Subtract the time-matched control from every dose
delta_df = paired_means.sub(paired_means[0], axis=0)

# Remove the control column (all differences are zero)
delta_df = delta_df.drop(columns=0)

# Convert back to a regular DataFrame
delta_df = delta_df.reset_index()

# Average the treatment effects across experiments
delta_summary = (
    delta_df
    .groupby("Time")[[2.5, 10.0, 20.0]]
    .agg(["mean", "std"])
)

display(delta_df.round(2))
display(delta_summary.round(2))
```

```python

def simulate_comparison(
    seed,
    true_effect=-100,
    n_experiments=3,
    n_images=3,
    n_cells=20
):
    rng = np.random.default_rng(seed)
    records = []

    for experiment in range(1, n_experiments + 1):
        experiment_variation = rng.normal(0, 50)

        for treatment in [0, 1]:

            for image in range(1, n_images + 1):
                image_variation = rng.normal(0, 30)

                for cell in range(1, n_cells + 1):
                    cell_variation = rng.normal(0, 80)

                    lifetime = (
                        1500
                        + experiment_variation
                        + image_variation
                        + cell_variation
                        + true_effect * treatment
                    )

                    records.append({
                        "Experiment": experiment,
                        "Treatment": treatment,
                        "Image": image,
                        "Cell": cell,
                        "Lifetime": lifetime
                    })

    return pd.DataFrame(records)
```

```python

def analyse_simulation(data):

    results = []

    # METHOD 1: Naive cell-level Welch t-test
    control = data.loc[data["Treatment"] == 0, "Lifetime"]
    treated = data.loc[data["Treatment"] == 1, "Lifetime"]

    naive = stats.ttest_ind(
        treated,
        control,
        equal_var=False
    )

    results.append({
        "Method": "Naive",
        "Effect": treated.mean() - control.mean(),
        "P_value": naive.pvalue
    })

    # METHOD 2: Experiment-level paired t-test
    experiment_means = (
        data.groupby(["Experiment", "Treatment"])["Lifetime"]
        .mean()
        .unstack()
    )

    differences = experiment_means[1] - experiment_means[0]

    paired = stats.ttest_1samp(differences, popmean=0)

    results.append({
        "Method": "Paired",
        "Effect": differences.mean(),
        "P_value": paired.pvalue
    })

    # METHOD 3: Image-level mixed-effects model
    image_means = (
        data.groupby(["Experiment", "Treatment", "Image"])
        .agg(Mean_Lifetime=("Lifetime", "mean"))
        .reset_index()
    )

    model = smf.mixedlm(
        "Mean_Lifetime ~ Treatment",
        data=image_means,
        groups=image_means["Experiment"]
    )

    fitted = model.fit(reml=True, disp=False)

    results.append({
        "Method": "Mixed",
        "Effect": fitted.params["Treatment"],
        "P_value": fitted.pvalues["Treatment"]
    })

    return pd.DataFrame(results)
```

```python

import warnings
from scipy import stats
from statsmodels.tools.sm_exceptions import ConvergenceWarning

def analyse_simulation(data):

    results = []

    # 1. Naive cell-level Welch test
    control = data.loc[data["Treatment"] == 0, "Lifetime"]
    treated = data.loc[data["Treatment"] == 1, "Lifetime"]

    naive = stats.ttest_ind(treated, control, equal_var=False)

    naive_ci = naive.confidence_interval(confidence_level=0.95)

    results.append({
        "Method": "Naive",
        "Effect": treated.mean() - control.mean(),
        "CI_Lower": naive_ci.low,
        "CI_Upper": naive_ci.high,
        "P_value": naive.pvalue,
        "Converged": True,
        "Boundary": False,
        "Warnings": ""
    })

    # 2. Experiment-level paired test
    experiment_means = (
        data.groupby(["Experiment", "Treatment"])["Lifetime"]
        .mean()
        .unstack()
    )

    differences = experiment_means[1] - experiment_means[0]
    paired = stats.ttest_1samp(differences, popmean=0)

    paired_ci = paired.confidence_interval(confidence_level=0.95)

    results.append({
        "Method": "Paired",
        "Effect": differences.mean(),
        "CI_Lower": paired_ci.low,
        "CI_Upper": paired_ci.high,
        "P_value": paired.pvalue,
        "Converged": True,
        "Boundary": False,
        "Warnings": ""
    })

    # 3. Image-level mixed-effects model
    image_means = (
        data.groupby(["Experiment", "Treatment", "Image"])
        .agg(Mean_Lifetime=("Lifetime", "mean"))
        .reset_index()
    )

    model = smf.mixedlm(
        "Mean_Lifetime ~ Treatment",
        data=image_means,
        groups=image_means["Experiment"]
    )

    try:
        with warnings.catch_warnings(record=True) as caught:
            warnings.simplefilter("always")
            fitted = model.fit(reml=True, disp=False)

        warning_text = "; ".join(
            str(w.message) for w in caught
        )

        ci = fitted.conf_int().loc["Treatment"]

        group_variance = float(fitted.cov_re.iloc[0, 0])

        results.append({
            "Method": "Mixed",
            "Effect": fitted.params["Treatment"],
            "CI_Lower": ci.iloc[0],
            "CI_Upper": ci.iloc[1],
            "P_value": fitted.pvalues["Treatment"],
            "Converged": bool(fitted.converged),
            "Boundary": group_variance < 1e-4,
            "Warnings": warning_text
        })

    except Exception as error:
        results.append({
            "Method": "Mixed",
            "Effect": np.nan,
            "CI_Lower": np.nan,
            "CI_Upper": np.nan,
            "P_value": np.nan,
            "Converged": False,
            "Boundary": np.nan,
            "Warnings": str(error)
        })

    return pd.DataFrame(results)
```

```python

def simulate_comparison_v2(
    seed,
    true_effect=-100,
    n_experiments=3,
    n_images=3,
    n_cells=20,
    experiment_sd=50,
    treatment_sd=40,
    image_sd=30,
    cell_sd=80
):

    rng = np.random.default_rng(seed)
    records = []

    for experiment in range(1, n_experiments + 1):

        experiment_offset = rng.normal(0, experiment_sd)

        # Each experiment has its own treatment response
        treatment_offset = rng.normal(0, treatment_sd)

        for treatment in [0, 1]:

            for image in range(1, n_images + 1):

                image_offset = rng.normal(0, image_sd)

                for cell in range(1, n_cells + 1):

                    lifetime = (
                        1500
                        + experiment_offset
                        + image_offset
                        + cell_sd * rng.normal()
                        + treatment * (true_effect + treatment_offset)
                    )

                    records.append({
                        "Experiment": experiment,
                        "Treatment": treatment,
                        "Image": image,
                        "Cell": cell,
                        "Lifetime": lifetime
                    })

    return pd.DataFrame(records)
```

```python

def analyse_v2(data):

    results = []

    image_means = (
        data.groupby(["Experiment", "Treatment", "Image"])
        .agg(Mean_Lifetime=("Lifetime", "mean"))
        .reset_index()
    )

    # Paired experiment-level analysis
    experiment_means = (
        image_means.groupby(["Experiment", "Treatment"])
        ["Mean_Lifetime"].mean().unstack()
    )

    differences = experiment_means[1] - experiment_means[0]
    paired = stats.ttest_1samp(differences, popmean=0)
    paired_ci = paired.confidence_interval()

    results.append({
        "Method": "Paired",
        "Effect": differences.mean(),
        "SE": stats.sem(differences),
        "CI_Lower": paired_ci.low,
        "CI_Upper": paired_ci.high,
        "P_value": paired.pvalue,
        "Converged": True,
        "Warnings": ""
    })

    # Two mixed-effects specifications
    model_specs = {
        "Random intercept": "1",
        "Random slope": "~Treatment"
    }

    for method_name, random_formula in model_specs.items():

        try:
            with warnings.catch_warnings(record=True) as caught:
                warnings.simplefilter("always")

                model = smf.mixedlm(
                    "Mean_Lifetime ~ Treatment",
                    data=image_means,
                    groups=image_means["Experiment"],
                    re_formula=random_formula
                )

                fitted = model.fit(
                    reml=True,
                    method="lbfgs",
                    disp=False
                )

            ci = fitted.conf_int().loc["Treatment"]

            results.append({
                "Method": method_name,
                "Effect": fitted.params["Treatment"],
                "SE": fitted.bse["Treatment"],
                "CI_Lower": ci.iloc[0],
                "CI_Upper": ci.iloc[1],
                "P_value": fitted.pvalues["Treatment"],
                "Converged": bool(fitted.converged),
                "Warnings": "; ".join(str(w.message) for w in caught)
            })

        except Exception as error:
            results.append({
                "Method": method_name,
                "Effect": np.nan,
                "SE": np.nan,
                "CI_Lower": np.nan,
                "CI_Upper": np.nan,
                "P_value": np.nan,
                "Converged": False,
                "Warnings": str(error)
            })

    return pd.DataFrame(results)
```

```python

def run_v2_simulations(
    n_simulations=30,
    experiment_counts=(3, 5, 10),
    true_effects=(-100, 0)
):

    all_results = []

    for n_experiments in experiment_counts:

        for true_effect in true_effects:

            for simulation in range(n_simulations):

                seed = (
                    700000
                    + n_experiments * 10000
                    + int(true_effect == 0) * 1000
                    + simulation
                )

                data = simulate_comparison_v2(
                    seed=seed,
                    true_effect=true_effect,
                    n_experiments=n_experiments,
                    n_images=3,
                    n_cells=20,
                    treatment_sd=40
                )

                results = analyse_v2(data)

                results["Simulation"] = simulation + 1
                results["N_Experiments"] = n_experiments
                results["True_Effect"] = true_effect

                all_results.append(results)

    return pd.concat(all_results, ignore_index=True)
```

```python

v2_simulation_results = run_v2_simulations(
    n_simulations=30
)

print("Total analysis records:", len(v2_simulation_results))

display(
    v2_simulation_results.groupby(
        ["N_Experiments", "True_Effect", "Method"]
    )["Converged"]
    .mean()
    .unstack("Method")
    .round(3)
)
```

```python

# Calculate statistical performance
v2 = v2_simulation_results.copy()

v2["Coverage"] = (
    (v2["CI_Lower"] <= v2["True_Effect"]) &
    (v2["CI_Upper"] >= v2["True_Effect"])
)

v2["Significant"] = v2["P_value"] < 0.05
v2["Bias"] = v2["Effect"] - v2["True_Effect"]

v2["Valid"] = (
    v2["Converged"]
    & v2[["Effect", "P_value", "CI_Lower", "CI_Upper"]]
      .notna().all(axis=1)
)

valid = v2[v2["Valid"]].copy()

summary = (
    valid.groupby(["N_Experiments", "True_Effect", "Method"])
    .agg(
        N=("Simulation", "count"),
        Bias=("Bias", "mean"),
        Coverage=("Coverage", "mean"),
        Rejection_Rate=("Significant", "mean")
    )
    .reset_index()
)

print("STATISTICAL PERFORMANCE")
display(summary.round(3))

print("\nMODEL CONVERGENCE")
display(
    v2.groupby(["N_Experiments", "True_Effect", "Method"])
    ["Valid"].mean().unstack("Method").round(3)
)

# Plot the null-effect scenario
null = summary[summary["True_Effect"] == 0]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.lineplot(
    data=null,
    x="N_Experiments",
    y="Rejection_Rate",
    hue="Method",
    marker="o",
    ax=axes[0]
)
axes[0].axhline(0.05, color="grey", linestyle="--")
axes[0].set_title("A | False-positive rate")
axes[0].set_ylabel("Proportion significant")

sns.lineplot(
    data=null,
    x="N_Experiments",
    y="Coverage",
    hue="Method",
    marker="o",
    legend=False,
    ax=axes[1]
)
axes[1].axhline(0.95, color="grey", linestyle="--")
axes[1].set_title("B | 95% CI coverage")
axes[1].set_ylabel("Coverage")

convergence = (
    v2.groupby(["N_Experiments", "True_Effect", "Method"])
    ["Valid"].mean().reset_index()
)

sns.lineplot(
    data=convergence[convergence["True_Effect"] == 0],
    x="N_Experiments",
    y="Valid",
    hue="Method",
    marker="o",
    legend=False,
    ax=axes[2]
)
axes[2].set_title("C | Successful fit rate")
axes[2].set_ylabel("Proportion valid")

for ax in axes:
    ax.set_xticks([3, 5, 10])
    ax.set_xlabel("Independent experiments")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```


# Conclusions

## Project overview

This project developed a reproducible Python workflow for analysing
synthetic fluorescence lifetime imaging microscopy (FLIM)-like data.

The simulated datasets contained a hierarchical experimental structure:

- Independent biological experiments
- Treatment and control conditions
- Multiple microscopy images per condition
- Multiple cells measured within each image

## Key findings

### 1. Pseudoreplication

Treating individual cells as independent observations produced
overly narrow confidence intervals and inflated false-positive rates.

### 2. Experimental design

Increasing the number of cells within existing images provided
limited improvements in treatment-effect precision compared with
collecting additional images or independent experiments under the
initial simulation assumptions.

### 3. Mixed-effects modelling

Random-intercept models accounted for experiment-level baseline
variation but could become unstable with few experiments.

### 4. Biological heterogeneity

Allowing treatment effects to vary between experiments revealed
limitations of random-intercept-only models.

Random-slope models represented this additional variability but
required sufficient biological replication for stable estimation.

### 5. Statistical uncertainty

The paired experiment-level approach provided a useful benchmark
for estimating treatment effects and uncertainty in balanced,
paired experimental designs.

## Limitations

- All data were simulated, not experimentally measured.
- Simulation parameters were illustrative rather than fitted to real data.
- Monte Carlo sample sizes were relatively small.
- Mixed-model inference relied on asymptotic approximations.
- The simulated experimental structure was balanced.
- Model performance may differ for other biological and measurement
  variability structures.

## Conclusion

Statistical reliability depends not only on the number of observations
collected, but also on the hierarchical structure of the experiment
and the assumptions used during analysis.
